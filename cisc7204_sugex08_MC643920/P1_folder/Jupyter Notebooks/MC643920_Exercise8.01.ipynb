## Exercise8.01 - 01. Read and inspect DataAccessObject

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('../../Support').resolve()))
import course_maps
import geoplotlib
import pandas as pd
import numpy as np
from geoplotlib.utils import DataAccessObject, BoundingBox, read_csv
from IPython.display import display
DATA = Path('../../Datasets')
print('LU MENGJIE | MC643920 | Suggested Exercise 08')
print('Public Esri gray basemaps replace key-protected CARTO; HOT is the custom provider.')
dataset = read_csv(str(DATA/'poaching_points_cleaned.csv'))
print(type(dataset))
print(dataset)
pd_dataset = pd.read_csv(DATA/'poaching_points_cleaned.csv')
display(pd_dataset.head())
print('DataAccessObject wraps arrays for plotting; pandas DataFrame adds tabular transformation and analysis.')
```

## Exercise8.01 - 02. Dot density

```python
geoplotlib.dot(dataset)
course_maps.snapshot('poaching_dot')
```

## Exercise8.01 - 03. Histogram, binsize 20

```python
geoplotlib.hist(dataset, binsize=20)
course_maps.snapshot('poaching_hist')
```

## Exercise8.01 - 04. Voronoi

```python
geoplotlib.voronoi(dataset, cmap='Blues_r', max_area=1e5, alpha=255)
course_maps.snapshot('poaching_voronoi')
```
