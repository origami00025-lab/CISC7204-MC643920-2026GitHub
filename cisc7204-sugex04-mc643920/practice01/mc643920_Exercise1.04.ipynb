# Exercise 1.04 - Loading a Sample Dataset and Calculating the Mean

This notebook follows the Pandas Basics tutorial. The sample file contains World Bank population-density values (people per square kilometre) by country and year. The first index column is the country name.

```python
import pandas as pd

# The submitted folder keeps Datasets beside the four practice folders.
dataset = pd.read_csv("../Datasets/world_population.csv", index_col=0)
numeric = dataset.select_dtypes(include="number")
```

```python
print("First five rows:")
dataset.head()
```

```python
print("DataFrame shape:", dataset.shape)
print("Index name:", dataset.index.name)
print("Number of numeric year columns:", len(numeric.columns))
```

```python
mean_1961 = numeric["1961"].mean()
mean_2015 = numeric["2015"].mean()
print("Mean population density in 1961:", mean_1961)
print("Mean population density in 2015:", mean_2015)
print("Change in the mean (2015 - 1961):", mean_2015 - mean_1961)
```

```python
print("Mean for each country - first 10:")
numeric.mean(axis=1).head(10)
```

```python
print("Mean for each year - last 10 year columns:")
numeric.mean(axis=0).tail(10)
```

```python
print("Mean of every numeric column in the complete dataset:")
numeric.mean()
```

The numeric table contains only the year columns, so text metadata such as the country code is not included in arithmetic.
