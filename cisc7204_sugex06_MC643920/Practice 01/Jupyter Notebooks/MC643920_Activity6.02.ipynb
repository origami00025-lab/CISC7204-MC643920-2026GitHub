**Purpose:** Compare Tomatometer and Audience Score percentages for the five movies, with major and minor y-axis intervals.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from pathlib import Path

def locate_datasets():
    here = Path.cwd().resolve()
    for folder in (here, *here.parents):
        candidate = folder / "Datasets"
        if candidate.is_dir():
            return candidate
    raise FileNotFoundError("Could not find the project Datasets folder.")

DATASETS = locate_datasets()
print("Datasets:", DATASETS)
```

```python
movies = pd.read_csv(DATASETS / 'movie_scores.csv', index_col=0)
movies
```

```python
positions = np.arange(len(movies))
bar_width = 0.34
fig, ax = plt.subplots(figsize=(10, 5), dpi=140)
ax.bar(positions - bar_width/2, movies['Tomatometer'], bar_width, label='Tomatometer')
ax.bar(positions + bar_width/2, movies['AudienceScore'], bar_width, label='Audience Score')
ax.set_xticks(positions, movies['MovieTitle'], rotation=20, ha='right')
ax.set_ylim(0, 105)
ax.set_yticks(np.arange(0, 101, 20))
ax.set_yticks(np.arange(0, 101, 5), minor=True)
ax.set_ylabel('Score (%)')
ax.set_title('Movie Scores Comparison')
ax.grid(axis='y', which='major', linewidth=0.8, alpha=0.55)
ax.grid(axis='y', which='minor', linewidth=0.35, alpha=0.25)
ax.legend()
fig.tight_layout()
plt.show()
```
