```python
from pathlib import Path
import pandas as pd
from IPython.display import display

source = Path('../Datasets') / '1. Weather Data.csv'
df = pd.read_csv(source)
print('Source:', source, '| rows:', len(df), '| columns:', len(df.columns))
display(df.head(3))
```

```python
df = df.rename(columns={'Weather': 'Weather Condition'})
print('Renamed column Weather → Weather Condition')
```

### Question 1.1: Unique wind speeds

```python
print('Unique wind speeds (km/h):', sorted(df['Wind Speed_km/h'].dropna().unique()))
print('Number of unique values:', df['Wind Speed_km/h'].nunique())
```

### Question 1.2: How many times was the weather exactly Clear?

```python
clear = df[df['Weather Condition'].eq('Clear')]
print('Clear records:', len(clear))
display(clear.head(10))
```

### Question 1.3: Wind speed equal to 4 km/h

```python
wind4 = df[df['Wind Speed_km/h'].eq(4)]
print('Records with wind speed 4 km/h:', len(wind4))
display(wind4.head(10))
```

### Question 1.4: Null values in each column

```python
display(df.isna().sum().rename('missing_values').to_frame())
```

### Question 1.5: Rename Weather to Weather Condition

```python
print('Column name after rename:', 'Weather Condition' if 'Weather Condition' in df.columns else 'NOT FOUND')
print('All columns:', df.columns.tolist())
```

### Question 1.6: Mean visibility

```python
print('Mean visibility (km):', df['Visibility_km'].mean())
```

### Question 1.7: Standard deviation of pressure

```python
print('Sample standard deviation of pressure (kPa):', df['Press_kPa'].std())
```

### Question 1.8: Variance of relative humidity

```python
print('Sample variance of relative humidity:', df['Rel Hum_%'].var())
```

### Question 1.9: All records containing Snow

```python
snow = df[df['Weather Condition'].str.contains('Snow', case=False, na=False)]
print('Snow-related records:', len(snow))
display(snow.head(12))
```

### Question 1.10: Wind speed above 24 km/h and visibility 25 km

```python
q = df[(df['Wind Speed_km/h'] > 24) & (df['Visibility_km'] == 25)]
print('Matching records:', len(q))
display(q.head(12))
```

### Question 1.11: Mean numeric values by weather condition

```python
display(df.groupby('Weather Condition').mean(numeric_only=True).round(2))
```

### Question 1.12: Minimum and maximum numeric values by weather condition

```python
extremes = df.groupby('Weather Condition').agg(['min', 'max'])
display(extremes)
```

### Question 1.13: Records where condition is exactly Fog

```python
fog = df[df['Weather Condition'].eq('Fog')]
print('Fog records:', len(fog))
display(fog.head(12))
```

### Question 1.14: Clear OR visibility above 40 km

```python
q = df[df['Weather Condition'].eq('Clear') | (df['Visibility_km'] > 40)]
print('Matching records:', len(q))
display(q.head(12))
```

### Question 1.15: (Clear AND humidity above 50%) OR visibility above 40 km

```python
q = df[(df['Weather Condition'].eq('Clear') & (df['Rel Hum_%'] > 50)) | (df['Visibility_km'] > 40)]
print('Matching records:', len(q))
display(q.head(12))
```
