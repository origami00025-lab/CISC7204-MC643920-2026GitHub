# Activity 1.02 - Forest Fire Size and Temperature Analysis

The forest-fire dataset records weather and fire observations from Montesinho Natural Park. This activity focuses on month, temp (degrees Celsius), and area (burned hectares).

```python
import pandas as pd

data = pd.read_csv("../Datasets/forestfires.csv")
print("First five rows:")
data.head()
```

```python
print("Rows where a non-zero area was burned:")
data[data["area"] > 0].head(10)
```

```python
area = data["area"]
print("Mean:", area.mean())
print("Min:", area.min())
print("Max:", area.max())
print("Standard deviation:", area.std())
print("Median:", area.median())
print("Mean greater than median:", area.mean() > area.median())
```

```python
print("Unique months:", sorted(data["month"].unique()))
print("Number of March records:", data[data["month"] == "mar"].shape[0])
```

```python
print("Monthly mean temperature and burned area:")
for month in sorted(data["month"].unique()):
    monthly = data[data["month"] == month]
    print(f"{month}: mean temp={monthly['temp'].mean():.2f}, mean area={monthly['area'].mean():.2f}")
```

```python
counts = data["month"].value_counts()
print("Month with the most fire records:", counts.idxmax())
print("Number of records in that month:", counts.max())
```

```python
monthly_summary = data.groupby("month")[["temp", "area"]].agg(["mean", "median", "count"])
monthly_summary
```
