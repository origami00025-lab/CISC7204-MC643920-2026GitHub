# Exercise 1.02 — Indexing, Slicing, Splitting, and Iterating

This notebook uses `normal_distribution_splittable.csv`, a 24 × 9 NumPy dataset.

```python
import numpy as np
from pathlib import Path

data_path = Path("..") / "Datasets" / "normal_distribution_splittable.csv"
dataset = np.genfromtxt(data_path, delimiter=",")
print("Dataset shape:", dataset.shape)
dataset
```

## Indexing

```python
# Mean of the second row
second_row = dataset[1]
np.mean(second_row)
```

```python
# Mean of the last row using reverse indexing
last_row = dataset[-1]
np.mean(last_row)
```

```python
# First value of the first row
first_value_first_row = dataset[0][0]
first_value_first_row
```

```python
# Last value of the second-last row
last_value_second_last_row = dataset[-2, -1]
last_value_second_last_row
```

## Slicing

```python
# 2 × 2 intersection beginning at the second row and second column
subsection_2x2 = dataset[1:3, 1:3]
subsection_2x2
```

```python
# Every other element of the fifth row
every_other_fifth_row = dataset[4, ::2]
every_other_fifth_row
```

```python
# Reverse the final row
reversed_last_row = dataset[-1, ::-1]
reversed_last_row
```

## Splitting

```python
# Split horizontally into three equal parts
horizontal_splits = np.hsplit(dataset, 3)
print([part.shape for part in horizontal_splits])
```

```python
# Split the first horizontal part vertically into two equal parts
vertical_splits = np.vsplit(horizontal_splits[0], 2)
print([part.shape for part in vertical_splits])
```

```python
# Compare the original and subset shapes
print("Dataset:", dataset.shape)
print("Subset:", vertical_splits[0].shape)
```

## Iterating

```python
# Iterate over every value; show a preview and the total count.
values = []
for value in np.nditer(dataset):
    values.append(float(value))
print("First 12 values:", values[:12])
print("Values visited:", len(values))
```

```python
# Iterate with row-and-column indices using ndenumerate
indexed_values = []
for index, value in np.ndenumerate(dataset):
    indexed_values.append((index, float(value)))
print("First 12 indexed values:", indexed_values[:12])
print("Indexed values visited:", len(indexed_values))
```
