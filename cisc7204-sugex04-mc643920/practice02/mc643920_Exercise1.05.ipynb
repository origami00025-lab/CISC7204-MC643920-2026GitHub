# Exercise 1.05 - Mean, Median, and Variance with Pandas

This notebook extends the previous exercise with descriptive statistics and demonstrates interoperability between Pandas and NumPy.

```python
import pandas as pd

# The submitted folder keeps Datasets beside the four practice folders.
dataset = pd.read_csv("../Datasets/world_population.csv", index_col=0)
numeric = dataset.select_dtypes(include="number")
```

```python
print("First two rows:")
dataset[0:2]
```

```python
print("Mean across the third row:")
dataset.iloc[[2]][numeric.columns].mean(axis=1)
```

```python
print("Mean across the last row:")
dataset.iloc[[-1]][numeric.columns].mean(axis=1)
```

```python
print("Mean across Germany's yearly values:")
dataset.loc[["Germany"]][numeric.columns].mean(axis=1)
```

```python
print("Median across the last row:")
dataset.iloc[[-1]][numeric.columns].median(axis=1)
```

```python
print("Median across the last three rows:")
dataset[-3:][numeric.columns].median(axis=1)
```

```python
print("Median across the first ten countries:")
dataset.head(10)[numeric.columns].median(axis=1)
```

```python
print("Variance for the last five years:")
numeric.var().tail()
```

```python
import numpy as np

pandas_mean = numeric["2015"].mean()
numpy_mean = np.mean(numeric["2015"])
print("Pandas mean for 2015:", pandas_mean)
print("NumPy mean for 2015:", numpy_mean)
print("Values agree:", np.isclose(pandas_mean, numpy_mean))
```
