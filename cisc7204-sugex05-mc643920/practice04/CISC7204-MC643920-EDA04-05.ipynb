```python
from pathlib import Path
import pandas as pd
from IPython.display import display

source = Path('../Datasets') / '4. Covid_19_data.csv'
df = pd.read_csv(source)
print('Source:', source, '| rows:', len(df), '| columns:', len(df.columns))
display(df.head(3))
```

### Question 4.1: Confirmed, deaths and recovered by region

```python
region_totals = df.groupby('Region')[['Confirmed','Deaths','Recovered']].sum()
display(region_totals)
```

### Question 4.2: Remove records with fewer than 10 confirmed cases

```python
before = len(df)
df = df[df['Confirmed'] >= 10].copy()
print('Removed:', before - len(df), 'Remaining:', len(df))
print('Minimum Confirmed remaining:', df['Confirmed'].min())
```

### Question 4.3: Region with the highest confirmed total

```python
region_totals = df.groupby('Region')['Confirmed'].sum()
maximum = region_totals.max()
display(region_totals[region_totals.eq(maximum)].to_frame('Confirmed_total'))
```

### Question 4.4: Region with the lowest death total

```python
death_totals = df.groupby('Region')['Deaths'].sum()
minimum = death_totals.min()
print('Minimum deaths:', minimum)
display(death_totals[death_totals.eq(minimum)].to_frame('Deaths_total'))
```

### Question 4.5: India totals as of 29 April 2020

```python
india = df[df['Region'].eq('India')]
print('Dates in India records:', india['Date'].unique().tolist())
display(india[['Confirmed','Deaths','Recovered']].sum().to_frame('India_total'))
```

### Question 4.6A: Sort records by Confirmed ascending

```python
confirmed_ascending = df.sort_values('Confirmed', ascending=True)
display(confirmed_ascending.head(12))
```

### Question 4.6B: Sort records by Recovered descending

```python
recovered_descending = df.sort_values('Recovered', ascending=False)
display(recovered_descending.head(12))
```
