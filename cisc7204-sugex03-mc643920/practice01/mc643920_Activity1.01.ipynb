# Activity 1.01 — Mean, Median, Variance, and Standard Deviation

This activity consolidates the NumPy operations from Exercise 1.01 using `normal_distribution.csv`.

```python
# Import NumPy and load the dataset
import numpy as np
from pathlib import Path

data_path = Path("..") / "Datasets" / "normal_distribution.csv"
dataset = np.genfromtxt(data_path, delimiter=",")
print("Dataset shape:", dataset.shape)
```

```python
# Print a subset of the first two rows
dataset[0:2]
```

## Mean

```python
# Mean of the third row
a_mean_third_row = np.mean(dataset[2])
a_mean_third_row
```

```python
# Mean of the last column
a_mean_last_column = np.mean(dataset[:, -1])
a_mean_last_column
```

```python
# Mean of the first three rows and first three columns
a_mean_submatrix = np.mean(dataset[0:3, 0:3])
a_mean_submatrix
```

## Median

```python
# Median of the last row
a_median_last_row = np.median(dataset[-1])
a_median_last_row
```

```python
# Median of the last three columns
a_median_last_columns = np.median(dataset[:, -3:])
a_median_last_columns
```

```python
# Median of each row
a_median_each_row = np.median(dataset, axis=1)
a_median_each_row
```

## Variance and standard deviation

```python
# Variance of each column
a_variance_each_column = np.var(dataset, axis=0)
a_variance_each_column
```

```python
# Variance of the last two rows and first two columns
a_variance_submatrix = np.var(dataset[-2:, :2])
a_variance_submatrix
```

```python
# Standard deviation for the complete dataset
a_standard_deviation = np.std(dataset)
a_standard_deviation
```
