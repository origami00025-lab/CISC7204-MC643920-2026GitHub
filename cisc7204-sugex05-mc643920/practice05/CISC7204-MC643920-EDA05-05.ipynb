```python
from pathlib import Path
import pandas as pd
from IPython.display import display

source = Path('../Datasets') / '5. London Housing Data (1).csv'
df = pd.read_csv(source)
print('Source:', source, '| rows:', len(df), '| columns:', len(df.columns))
display(df.head(3))
```

### Question 5.1: Convert date to datetime

```python
df['date'] = pd.to_datetime(df['date'], format='%m/%d/%Y')
print('date dtype:', df['date'].dtype)
display(df[['date','area','average_price']].head())
```

### Question 5.2A: Add a year column

```python
df['year'] = df['date'].dt.year
display(df[['date','year','area']].head())
```

### Question 5.2B: Insert month as the second column

```python
df.insert(1, 'month', df['date'].dt.month)
print('Column order:', df.columns.tolist())
display(df.head())
```

### Question 5.3: Remove year and month columns

```python
df = df.drop(columns=['year', 'month'])
print('Column order after removal:', df.columns.tolist())
```

### Question 5.4: Crime count equal to zero

```python
zero_crimes = df[df['no_of_crimes'].eq(0)]
print('Records where no_of_crimes is exactly 0:', len(zero_crimes))
display(zero_crimes.head(12))
```

### Question 5.5: Yearly minimum and maximum average price in England

```python
england = df[df['area'].str.casefold().eq('england')].copy()
england['year'] = england['date'].dt.year
print('England records:', len(england))
display(england.groupby('year')['average_price'].agg(['min','max']))
```

### Question 5.6: Minimum and maximum crimes by area

```python
display(df.groupby('area')['no_of_crimes'].agg(['min','max']))
```

### Question 5.7: Records per area with average price below 100000

```python
low_price = df[df['average_price'] < 100000]
print('Total matching records:', len(low_price))
display(low_price['area'].value_counts().rename_axis('area').to_frame('records'))
```
