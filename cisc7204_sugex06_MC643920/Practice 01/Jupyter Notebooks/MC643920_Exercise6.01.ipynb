**Purpose:** Create the tutorial line plot from four (x, y) pairs, using circle markers joined by line segments.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Plot the four points as circles connected by line segments.
x = [1, 2, 4, 5]
y = [1, 3, 4, 3]

fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
ax.plot(x, y, 'o-')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('A Simple Visualization')
fig.tight_layout()
plt.show()
```
