<h1> Task #3: </h1>
<h3>Replace '?' with 'NaN'</h3>
Replace the '?' entries in the dataset with NaN value, recevied from the Numpy package.


````python
df.replace('?', np.nan, inplace=True)
print('Missing values by column after replacement:')
print(df.isna().sum())
````

**Explanation:** The replacement step converts the dataset's `?` markers to `NaN`; the printed counts provide visible evidence of the resulting missing values, answering Task 3.
