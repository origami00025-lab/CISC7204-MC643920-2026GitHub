# Exercise 1.06 - Indexing, Slicing, and Iterating with Pandas

This notebook practices label-based indexing (loc), position-based indexing (iloc), slices, and row iteration on the population-density table.

```python
import pandas as pd

# The submitted folder keeps Datasets beside the four practice folders.
dataset = pd.read_csv("../Datasets/world_population.csv", index_col=0)
numeric = dataset.select_dtypes(include="number")
```

```python
print("United States - first five columns:")
dataset.loc[["United States"]].iloc[:, :5]
```

```python
print("Second-to-last row:")
dataset.iloc[[-2]]
```

```python
print("Year 2000 - first five countries:")
dataset["2000"].head()
```

```python
print("India in year 2000 as a one-row DataFrame:")
dataset[["2000"]].loc[["India"]]
```

```python
india_2000 = numeric["2000"].loc["India"]
print("India's 2000 population density:", india_2000)
```

```python
print("Rows 2 through 5 by position:")
dataset.iloc[1:5]
```

```python
country_list = ["Germany", "Singapore", "United States", "India"]
year_list = ["1970", "1990", "2010"]
dataset.loc[country_list, year_list]
```

```python
print("Selected values for the four countries and three years:")
dataset.loc[country_list, year_list].to_numpy()
```

```python
print("Countries before Angola:")
for index, row in dataset.iterrows():
    if index == "Angola":
        break
    print(index, row[["Country Code", "1970", "1990", "2010"]].to_dict())
```
