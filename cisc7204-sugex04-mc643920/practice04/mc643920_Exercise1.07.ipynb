# Exercise 1.07 - Filtering, Sorting, and Reshaping

This notebook applies advanced Pandas operations to the same population-density dataset: filtering rows and columns, sorting values, and reshaping a 2015 snapshot.

```python
import pandas as pd

# The submitted folder keeps Datasets beside the four practice folders.
dataset = pd.read_csv("../Datasets/world_population.csv", index_col=0)
numeric = dataset.select_dtypes(include="number")
```

```python
print("Selected year columns:")
dataset.filter(items=["1961", "2000", "2015"]).head()
```

```python
print("Countries with density above 500 in 2000:")
dataset.loc[dataset["2000"] > 500, ["2000"]].head(15)
```

```python
print("Year columns beginning with 200:")
dataset.filter(regex=r"^200", axis=1).head()
```

```python
print("Rows whose country name starts with A:")
dataset.filter(regex=r"^A", axis=0).head()
```

```python
print("Rows containing 'land' in the country name:")
dataset.filter(like="land", axis=0).head()
```

```python
print("Ten smallest 1961 values:")
dataset.sort_values(by=["1961"])[["1961"]].head(10)
```

```python
print("Ten smallest 2015 values:")
dataset.sort_values(by=["2015"])[["2015"]].head(10)
```

```python
print("Ten largest 2015 values:")
dataset.sort_values(by=["2015"], ascending=False)[["2015"]].head(10)
```

```python
dataset_2015 = dataset[["Country Code", "2015"]].copy()
dataset_2015["Year"] = 2015
reshaped = dataset_2015.pivot(index="Year", columns="Country Code", values="2015")
print("One-row table with country codes as columns:")
reshaped.iloc[:, :12]
```
