## Exercise7.03 — 01. Import Squarify and read water usage

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from IPython.display import display

DATA_DIR = Path('../../Datasets')
pd.set_option('display.max_columns', 8)
pd.set_option('display.width', 110)
print('LU MENGJIE | MC643920 | Python 3 / Seaborn', sns.__version__)
import squarify
mydata = pd.read_csv(DATA_DIR / 'water_usage.csv', index_col=0)
print('Usage categories:', len(mydata))
print('Percentage total:', mydata['Percentage'].sum())
display(mydata)
```

## Exercise7.03 — 02. Create labels and the Seaborn light palette

```python
labels = mydata['Usage'] + ' (' + mydata['Percentage'].astype(str) + '%)'
colors = sns.light_palette('green', mydata.shape[0])
print('Treemap labels:')
for label in labels:
    print(label)
print('Palette colors:', len(colors))
```

## Exercise7.03 — 03. Draw the Squarify treemap

```python
plt.figure(figsize=(8, 4.8), dpi=110)
squarify.plot(sizes=mydata['Percentage'], label=labels, color=colors,
              text_kwargs={'fontsize': 12})
plt.axis('off')
plt.title('Water usage')
plt.tight_layout()
plt.show()
```

Toilet use is the largest share at 24%, followed by showers at 20% and faucets at 19%. Leaks account for 12%. The rectangle areas represent shares; the palette supplies visual separation rather than a second numerical measure.
