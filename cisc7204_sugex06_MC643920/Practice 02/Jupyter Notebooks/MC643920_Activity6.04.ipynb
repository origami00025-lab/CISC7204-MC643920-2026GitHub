**Purpose:** Read the quarterly smartphone-sales dataset and compare unit sales for five manufacturers with a stacked area chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from pathlib import Path

def locate_datasets():
    here = Path.cwd().resolve()
    for folder in (here, *here.parents):
        candidate = folder / "Datasets"
        if candidate.is_dir():
            return candidate
    raise FileNotFoundError("Could not find the project Datasets folder.")

DATASETS = locate_datasets()
print("Datasets:", DATASETS)
```

```python
sales = pd.read_csv(DATASETS / 'smartphone_sales.csv', index_col=0)
sales
```

```python
manufacturers = ['Apple', 'Samsung', 'Huawei', 'Xiaomi', 'OPPO']
fig, ax = plt.subplots(figsize=(9, 5), dpi=140)
ax.stackplot(sales['Quarter'], *(sales[name] for name in manufacturers), labels=manufacturers)
ax.set_title('Smartphone Sales Units')
ax.set_xlabel('Quarter')
ax.set_ylabel('Sales units')
ax.legend(loc='upper left', ncol=2)
ax.grid(axis='y', alpha=0.3)
fig.tight_layout()
plt.show()
```
