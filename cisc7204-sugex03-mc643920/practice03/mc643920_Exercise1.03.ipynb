# Exercise 1.03 — Filtering, Sorting, Combining, and Reshaping

This notebook applies advanced NumPy operations to `normal_distribution_splittable.csv`.

```python
import numpy as np
from pathlib import Path

data_path = Path("..") / "Datasets" / "normal_distribution_splittable.csv"
dataset = np.genfromtxt(data_path, delimiter=",")
print("Dataset shape:", dataset.shape)
dataset
```

## Filtering

```python
# Values greater than 105
values_greater_105 = dataset[dataset > 105]
print("Count:", values_greater_105.size)
values_greater_105
```

```python
# Values strictly between 90 and 95
values_between_90_95 = np.extract((dataset > 90) & (dataset < 95), dataset)
print("Count:", values_between_90_95.size)
values_between_90_95
```

```python
# Indices of values whose distance from 100 is less than 1
rows, columns = np.where(np.abs(dataset - 100) < 1)
one_away_indices = list(zip(rows.tolist(), columns.tolist()))
print("Matching positions:", len(one_away_indices))
one_away_indices
```

## Sorting

```python
# Sort each row (the last axis by default)
row_sorted = np.sort(dataset)
row_sorted
```

```python
# Sort values by column
column_sorted = np.sort(dataset, axis=0)
column_sorted
```

```python
# Use argsort and fancy indexing to sort the first row
index_sorted = np.argsort(dataset[0])
sorted_first_row = dataset[0][index_sorted]
print("Sorted indices:", index_sorted)
sorted_first_row
```

## Combining

```python
# Split into thirds, then split the first third into two row blocks
thirds = np.hsplit(dataset, 3)
halved_first = np.vsplit(thirds[0], 2)
print("Third shapes:", [part.shape for part in thirds])
print("First-third halves:", [part.shape for part in halved_first])
```

```python
# Vertically combine the two halves of the first third
first_column_block = np.vstack([halved_first[0], halved_first[1]])
first_column_block
```

```python
# Add the second third horizontally
first_second_block = np.hstack([first_column_block, thirds[1]])
print("Shape after adding second block:", first_second_block.shape)
first_second_block
```

```python
# Add the final third horizontally to reconstruct the full dataset
full_data = np.hstack([first_second_block, thirds[2]])
print("Reconstructed shape:", full_data.shape)
print("Reconstruction matches original:", np.array_equal(full_data, dataset))
```

## Reshaping

```python
# Reshape the dataset into one row
single_list = np.reshape(dataset, (1, -1))
print("Shape:", single_list.shape)
single_list
```

```python
# Reshape the dataset into two columns; NumPy infers the row count
two_column_dataset = dataset.reshape(-1, 2)
print("Shape:", two_column_dataset.shape)
two_column_dataset
```
