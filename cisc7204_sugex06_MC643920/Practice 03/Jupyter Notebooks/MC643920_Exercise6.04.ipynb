**Purpose:** Build four polar charts for the employee ratings across Efficiency, Quality, Commitment, Responsible Conduct, and Cooperation.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
evaluations = pd.DataFrame({
    'Employee': ['Alex', 'Alice', 'Chris', 'Jennifer'],
    'Efficiency': [5, 4, 4, 3],
    'Quality': [5, 5, 3, 3],
    'Commitment': [5, 4, 4, 4],
    'Responsible Conduct': [4, 4, 4, 3],
    'Cooperation': [4, 4, 5, 5],
})
evaluations
```

```python
attributes = list(evaluations.columns[1:])
values = evaluations[attributes].to_numpy()
angles = np.linspace(0, 2*np.pi, len(attributes), endpoint=False)
angles = np.r_[angles, angles[0]]
values = np.c_[values, values[:, 0]]

fig, axes = plt.subplots(2, 2, figsize=(7, 7), dpi=125, subplot_kw={'polar': True})
for i, ax in enumerate(axes.flat):
    ax.plot(angles, values[i], linewidth=1.6)
    ax.set_ylim(0, 5)
    ax.set_yticks([1, 2, 3, 4, 5])
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(attributes, fontsize=8)
    ax.set_title(evaluations.loc[i, 'Employee'], fontsize=12, color='red', pad=12)
fig.tight_layout()
plt.show()
```
