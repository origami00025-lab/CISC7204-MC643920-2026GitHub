```python
fig, ax = plt.subplots(figsize=(6, 4), dpi=140)
ax.hist(iq_scores, bins=10)
ax.axvline(100, color='red', label='Mean = 100')
ax.axvline(115, color='red', linestyle='--', label='±1 SD')
ax.axvline(85, color='red', linestyle='--')
ax.set_xlabel('IQ score')
ax.set_ylabel('Frequency')
ax.set_title('IQ Scores for a Test Group of 100 Adults')
ax.legend()
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 4), dpi=140)
ax.boxplot(iq_scores)
ax.set_xticks([1], ['Test group'])
ax.set_ylabel('IQ score')
ax.set_title('IQ Scores for a Test Group of 100 Adults')
fig.tight_layout()
plt.show()
```
