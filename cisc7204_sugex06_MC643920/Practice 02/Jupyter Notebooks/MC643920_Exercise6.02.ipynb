```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from pathlib import Path

def locate_datasets():
    here = Path.cwd().resolve()
    for folder in (here, *here.parents):
        candidate = folder / "Datasets"
        if candidate.is_dir():
            return candidate
    raise FileNotFoundError("Could not find the project Datasets folder.")

DATASETS = locate_datasets()
print("Datasets:", DATASETS)
```

```python
water = pd.read_csv(DATASETS / 'water_usage.csv', index_col=0)
water
```

```python
fig, ax = plt.subplots(figsize=(7, 7), dpi=150)
explode = (0, 0, 0.1, 0, 0, 0)
ax.pie(water['Percentage'], explode=explode, labels=water['Usage'], autopct='%.0f%%')
ax.set_title('Water Usage')
fig.tight_layout()
plt.show()
```
