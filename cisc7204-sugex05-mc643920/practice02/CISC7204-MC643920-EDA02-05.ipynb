```python
from pathlib import Path
import pandas as pd
from IPython.display import display

source = Path('../Datasets') / '2. Cars Data.csv'
df = pd.read_csv(source)
print('Source:', source, '| rows:', len(df), '| columns:', len(df.columns))
display(df.head(3))
```

### Question 2.1: Find and fill numeric missing values

```python
print('Missing values before:')
display(df.isna().sum().to_frame('before'))
num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
print('Missing values after numeric mean imputation:')
display(df.isna().sum().to_frame('after'))
```

### Question 2.2: Counts of each car Make

```python
display(df['Make'].value_counts().rename_axis('Make').to_frame('count'))
```

### Question 2.3: Cars from Asia or Europe

```python
asia_europe = df[df['Origin'].isin(['Asia', 'Europe'])]
print('Asia or Europe records:', len(asia_europe))
display(asia_europe.head(12))
```

### Question 2.4: Remove cars weighing more than 4000

```python
before = len(df)
df = df[df['Weight'] <= 4000].copy()
print('Removed:', before - len(df), 'Remaining:', len(df))
print('Maximum remaining weight:', df['Weight'].max())
```

### Question 2.5: Increase city MPG by 3 for remaining cars

```python
df['MPG_City'] = df['MPG_City'] + 3
print('Updated MPG_City:')
display(df[['Make', 'Model', 'MPG_City']].head(12))
```
